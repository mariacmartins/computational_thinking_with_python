# Manipulação de Arquivos (txt) com Python

````python
# Escrevendo arquivos
arquivo = open("alunos.txt", "w", encoding="utf-8")

arquivo.write("Maria")
arquivo.write("João")

arquivo.close()
````

````python
# Escrevendo arquivos pulando linhas
arquivo = open("alunos.txt", "w", encoding="utf-8")

arquivo.write("Maria\n")
arquivo.write("João\n")
arquivo.write("Ana\n")

arquivo.close()
````

````python
# Lendo arquivos
arquivo = open("alunos.txt", "r", encoding="utf-8")

conteudo = arquivo.read()

print(conteudo)

arquivo.close()
````

````python
type(conteudo)
````

````python
# Lendo linha por linha

arquivo = open("alunos.txt", "r", encoding="utf-8")

for linha in arquivo:
    print(linha)

arquivo.close()
````

read() é interessante quando quero trabalhar com o conteúdo inteiro.
Percorrer o arquivo linha a linha é interessante quando quero processar cada registro individualmente.

````python
# Usando comando with
with open("alunos.txt", "r", encoding="utf-8") as arquivo:
    conteudo = arquivo.read()

print(conteudo)
````

Com o comando with, o Python cuida do fechamento do arquivo. Quando o bloco termina, o arquivo é fechado automaticamente.

````python
with open("resultado.txt", "w", encoding="utf-8") as arquivo:
    arquivo.write("Primeira linha\n")
    arquivo.write("Segunda linha\n")
    arquivo.write("Terceira linha\n")
````

````python
with open("resultado.txt", "w", encoding="utf-8") as arquivo:
    arquivo.write("Primeira linha\nSegunda linha\n")
````

````python
with open("log.txt", "a", encoding="utf-8") as arquivo:
    arquivo.write("Novo processamento\n")
````

````python
# Não substituirá o conteudo como o w, só adicionará ao final do arquivo
with open("log.txt", "a", encoding="utf-8") as arquivo:
    arquivo.write("Novo processamento - teste 2")
````

````python
# Transformando uma lista em um txt
alunos = [
    "Maria",
    "João",
    "Ana",
    "Carlos"
]

with open("alunos.txt", "w", encoding="utf-8") as arquivo:
    for aluno in alunos:
        arquivo.write(aluno + "\n")
````

````python
# Lendo arquivos e transformando dados
with open("alunos.txt", "r", encoding="utf-8") as arquivo:
    for linha in arquivo:
        nome = linha.strip().upper()
        print(nome)
````

````python
# Criando um novo arquivo a partir de outro
with open("alunos.txt", "r", encoding="utf-8") as entrada:
    with open("alunos_uppercase.txt", "w", encoding="utf-8") as saida:

        for linha in entrada:
            nome = linha.strip().upper()
            saida.write(nome + "\n")
````

### Comando readline() — lendo um arquivo linha por linha

O método readline() serve para ler uma única linha de um arquivo por vez.

````python
with open("alunos_uppercase.txt", "r", encoding="utf-8") as arquivo:

    linha = arquivo.readline()

    print(linha)
````

O readline() leu somente a primeira linha.

````python
with open("alunos_uppercase.txt", "r", encoding="utf-8") as arquivo:

    linha1 = arquivo.readline()
    linha2 = arquivo.readline()
    linha3 = arquivo.readline()

    print(linha1)
    print(linha2)
    print(linha3)
````

O \n representa a quebra de linha.

Por isso é muito comum vermos:

````python
with open("alunos_uppercase.txt", "r", encoding="utf-8") as arquivo:

    linha = arquivo.readline().strip()

    print(linha)
````

readline() + for

Essa combinação é especialmente interessante para arquivos com cabeçalho.

````python
with open("vendas.txt", "w", encoding="utf-8") as arquivo:
    arquivo.write(
        "produto;quantidade;preco\n"
        "Notebook;2;3500\n"
        "Mouse;5;80\n"
        "Teclado;3;150\n"
    )
````

````python
# Se quisermos ignorar o cabeçalho produto;quantidade;preco

with open("vendas.txt", "r", encoding="utf-8") as arquivo:

    arquivo.readline()

    for linha in arquivo:
        print(linha.strip())
````

lê a primeira linha e move a posição para a segunda.

Então o for começa aqui:

produto;quantidade;preco

↓ posição atual

Notebook;2;3500
Mouse;5;80
Teclado;3;150

readline() não existe para "pular cabeçalho". Ele existe para ler uma linha. Nós podemos aproveitar seu comportamento para ler o cabeçalho e simplesmente não armazená-lo.

````python
# Se quisermos guardar o cabeçalho
with open("vendas.txt", "r", encoding="utf-8") as arquivo:

    header = arquivo.readline().strip()

    print(f"Header: {header}")

    for linha in arquivo:
        print(linha.strip())
````
