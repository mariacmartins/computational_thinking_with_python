### Desafio Complementar

Crie um programa que solicite ao usuário 10 números inteiros e armazene os valores em um arquivo de texto chamado numeros.txt.

Cada número deve ser armazenado em uma linha diferente.

- Durante a entrada dos dados:

caso o usuário digite um valor que não seja um número inteiro, o programa deve utilizar tratamento de exceções para lidar com o erro;
o programa deve continuar solicitando um número até que consiga armazenar 10 números inteiros válidos.

- Depois de criar o arquivo:

abra o arquivo novamente para leitura;
leia os números armazenados;
calcule o somatório de todos os números;
mostre o resultado na tela.

Exemplo. Se o usuário informar:

```
10
5
8
3
12
7
4
20
6
15

```



O arquivo numeros.txt deverá ficar:

10
5
8
3
12
7
4
20
6
15

E o programa deverá apresentar:

Somatório: 90

````python
with open("numeros.txt", "w", encoding="utf-8") as arquivo:

    quantidade = 0

    while quantidade < 10:

        try:
            numero = int(input("Digite um número inteiro: "))

            arquivo.write(f"{numero}\n")

            quantidade += 1

        except ValueError:
            print("Digite um número inteiro válido.")
````

````python
soma = 0

with open("numeros.txt", "r", encoding="utf-8") as arquivo:

    for linha in arquivo:
        numero = int(linha.strip())

        soma += numero

print(f"Somatório: {soma}")
````
