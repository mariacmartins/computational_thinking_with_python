### Criando base tratada menor

```python
# Criando uma base tratada menor
df_mini = df[
    [
        "1.a_idade",
        "1.a.1_faixa_idade",
        "1.b_genero",
        "1.i.1_uf_onde_mora",
        "1.l_nivel_de_ensino",
        "4.a.1_atuacao_em_dados",
        "4.d.1_SQL",
        "4.d.2_R",
        "4.d.3_Python"
    ]
]
```

```python
df_mini.head()
```

```python
# Renomeando colunas
df_mini = df_mini.rename(
    columns={
        "1.a_idade": "idade",
        "1.a.1_faixa_idade": "faixa_etaria",
        "1.b_genero": "genero",
        "1.i.1_uf_onde_mora": "uf",
        "1.l_nivel_de_ensino": "escolaridade",
        "4.a.1_atuacao_em_dados": "atuacao",
        "4.d.1_SQL": "sql",
        "4.d.2_R": "r",
        "4.d.3_Python": "python"
    }
)

df_mini.head()
```

```python
faixas = (
    df_mini["faixa_etaria"]
    .value_counts()
)

faixas.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Distribuição dos respondentes por faixa etária")
plt.xlabel("Faixa etária")
plt.ylabel("Quantidade de respondentes")
plt.xticks(rotation=0)
plt.show()
```

### Transformando o csv em excel


```python
df_mini.to_excel(
    "state_of_data_tratado.xlsx",
    index=False
)
```

df_mini → DataFrame que queremos exportar

.to_excel() → salva o DataFrame no formato Excel

"state_of_data_tratado.xlsx" → nome do arquivo que será criado

index=False → não salva o índice do DataFrame como uma coluna no Excel

### Criando mini dataframes para exportar em abas

```python
df_rs = df[
    [
        "1.a_idade",
        "1.b_genero",
        "1.i.1_uf_onde_mora",
        "1.l_nivel_de_ensino",
        "4.a.1_atuacao_em_dados",
        "4.d.1_SQL",
        "4.d.3_Python"
    ]
]

df_rs = df_rs[
    df_rs["1.i.1_uf_onde_mora"] == "RS"
]

df_rs = df_rs.rename(
    columns={
        "1.a_idade": "idade",
        "1.b_genero": "genero",
        "1.i.1_uf_onde_mora": "uf",
        "1.l_nivel_de_ensino": "escolaridade",
        "4.a.1_atuacao_em_dados": "atuacao",
        "4.d.1_SQL": "usa_sql",
        "4.d.3_Python": "usa_python"
    }
)

df_rs.head()
```

```python
df_graduacao = df[
    [
        "1.a_idade",
        "1.b_genero",
        "1.i.1_uf_onde_mora",
        "1.l_nivel_de_ensino",
        "4.a.1_atuacao_em_dados",
        "4.d.1_SQL",
        "4.d.3_Python"
    ]
]

df_graduacao = df_graduacao[
    df_graduacao["1.l_nivel_de_ensino"] == "Estudante de Graduação"
]

df_graduacao = df_graduacao.rename(
    columns={
        "1.a_idade": "idade",
        "1.b_genero": "genero",
        "1.i.1_uf_onde_mora": "uf",
        "1.l_nivel_de_ensino": "escolaridade",
        "4.a.1_atuacao_em_dados": "atuacao",
        "4.d.1_SQL": "usa_sql",
        "4.d.3_Python": "usa_python"
    }
)

df_graduacao.head()
```

```python
# quantas linhas
print("Pessoas do RS:", len(df_rs))
print("Estudantes de graduação:", len(df_graduacao))
```

```python
# Exportar para diferentes abas do Excel
with pd.ExcelWriter("analises_state_of_data.xlsx") as writer:

    df_rs.to_excel(
        writer,
        sheet_name="Pessoas do RS",
        index=False
    )

    df_graduacao.to_excel(
        writer,
        sheet_name="Estudantes graduação",
        index=False
    )
```

### Comparando arquivos

```python
df.to_csv("state_of_data.csv", index=False)

df.to_excel("state_of_data.xlsx", index=False)

df.to_parquet("state_of_data.parquet", index=False)
```

```python
import os

tamanho_csv = os.path.getsize("state_of_data.csv")
tamanho_xlsx = os.path.getsize("state_of_data.xlsx")
tamanho_parquet = os.path.getsize("state_of_data.parquet")
```

```python
tamanho_csv_mb = tamanho_csv / (1024 ** 2)
tamanho_xlsx_mb = tamanho_xlsx / (1024 ** 2)
tamanho_parquet_mb = tamanho_parquet / (1024 ** 2)
```

```python
print(f"CSV:      {tamanho_csv_mb:.2f} MB")
print(f"Excel:    {tamanho_xlsx_mb:.2f} MB")
print(f"Parquet:  {tamanho_parquet_mb:.2f} MB")
```
