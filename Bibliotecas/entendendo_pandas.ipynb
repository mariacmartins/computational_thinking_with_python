### Entendendo pandas

```python
# Importando bibliotecas
import pandas as pd
import numpy as np
```

```python
# 1. Ler o arquivo CSV
df_jogadores = pd.read_csv("players_metrics.csv")

# 2. Exibir as primeiras linhas da tabela para entender as colunas
df_jogadores.head()
```

Ao rodar o comando pd.read_csv, o Pandas faz a leitura do arquivo de texto e o converte (carrega) na memória do computador estruturado como um DataFrame.

Tecnicamente falando, um Pandas DataFrame é uma estrutura de dados bidimensional, tabular e mutável em tamanho, com eixos rotulados (linhas e colunas).

```python
df_jogadores.describe()
```

```python
# Acessando uma coluna
df_jogadores['Nome']
```

```python
# Filtrando com a função query
df_brasil = df_jogadores.query("País == 'Brasil'")

df_brasil.head()
```

```python
# Se eu precisar filtrar baseado em uma variável
country = 'Alemanha'

df_alemanha = df_jogadores.query("País == @country")
df_alemanha.head()
```

```python
# 1. Isolamos as notas em uma Series do Pandas (em vez de uma lista comum)
notas_pandas = df_jogadores[['Ritmo', 'Finalização', 'Passe', 'Drible', 'Defesa', 'Físico']]

# 2. Calculamos direto com os métodos do Pandas
media_geral = notas_pandas.mean()
maior_nota = notas_pandas.max()

print(media_geral)
print(maior_nota)
```

```python
# Gerando gráficos
import seaborn as sns
import matplotlib.pyplot as plt

# 2. Define o tema visual do Seaborn (deixa o fundo bonito e com linhas de grade)
sns.set_theme(style="whitegrid")

# 3. Cria a figura do gráfico (largura x altura)
plt.figure(figsize=(10, 6))

# 4. O GRÁFICO DO SEABORN (regplot cria os pontos + a linha de tendência)
sns.regplot(
    data=df_jogadores,
    x="Idade",
    y="Ritmo"
)

# 5. Títulos e Legendas de forma simples
plt.title("Análise Física: Ritmo vs Idade dos Atletas", fontsize=16, fontweight="bold", pad=15)
plt.xlabel("Idade (Anos)", fontsize=12)
plt.ylabel("Ritmo (Velocidade)", fontsize=12)

# 6. Exibe o gráfico na tela
plt.show()
```

```python
df_filtrado_ordenado = df_jogadores.query("País == 'Brasil'").sort_values(by="Idade", ascending=False)

# CONFIGURANDO O VISUAL DO GRÁFICO
sns.set_theme(style="whitegrid") # Fundo limpo com linhas de grade suaves
plt.figure(figsize=(10, 5))      # Define o tamanho da janela do gráfico

#  CRIANDO O GRÁFICO DE BARRAS HORIZONTAIS
# Invertemos o X e o Y para a barra deitar: X vira o número (Idade) e Y vira o texto (Nome)
ax = sns.barplot(
    data=df_filtrado_ordenado,
    x="Idade",
    y="Nome",
    color="#1f77b4" # Um azul sólido e profissional
)

# TÍTULOS E AJUSTES FINAIS
plt.title("Ranking de Idade dos Atletas", fontsize=16, fontweight="bold", pad=15)
plt.xlabel("Idade (Anos)", fontsize=12)
plt.ylabel("Jogador", fontsize=12)

# Limpa as bordas do gráfico para um visual mais moderno
sns.despine(left=True, bottom=True)

# EXIBE O GRÁFICO
plt.tight_layout() # Ajusta o espaçamento para não cortar os nomes
plt.show()
```

### Tratando dados inconsistentes


```python
dados_streaming = {
    "ID_Usuario": [101, 102, 103, 102, 104, 105, 106, 107, None, 108, 109, 110, 111],
    "Nome": [
        "Ana Silva",
        " Carlos Souza",
        "Beatriz Costa",
        " Carlos Souza",
        "  Diego Lima  ",
        "Mariana Reis",
        "Ricardo Alves",
        "Juliana Mendes",
        "Usuario Anonimo",
        "Fernando Pessoa",
        "Roberto Carlos",
        "Amanda Melo",
        "Pedro Albuquerque"
    ],
    "Plano": ["Premium", "Free", "Medium", "Free", "free ", "PREMIUM", "Premium", None, "Free", "Premium", "Free", "Premium", "Free"],
    "Idade": [28, 34, 22, 34, 19, 29, 45, 31, 25, "vinte e seis", 37.5, -5, 120],
    "Minutos_Assistidos": [1240, 150, 890, 150, 420, 2100, None, 680, 10, 530, 310, 1500, 5],
    "Cancelou": ["Não", "Sim", "Não", "Sim", "Não", "Não", "Não", "Não", "Sim", "Não", "Não", "Não", "Sim"]
}

# Transformando o Dicionário em um DataFrame do Pandas
df = pd.DataFrame(dados_streaming)

df.head(10)
```

```python
# Comando A: Exibe o formato da tabela (Linhas, Colunas)
print(f"Formato inicial do DataFrame: {df.shape} (Linhas, Colunas)\n")
```

```python
df.info()
```

```python
# Checando dados duplicados
df.duplicated().sum()
```

```python
df[df.duplicated()]
```

```python
# Checando valores nulos
df.isna().sum()
```

```python
# Tratando duplicados e nulos críticos
df = df.drop_duplicates()
df = df.dropna(subset=["ID_Usuario"])
```

```python
# Convertendo tipos
df["ID_Usuario"] = df["ID_Usuario"].astype("int64")

df.head(3)
```

```python
df['Idade']
```

```python
df["Idade"] = pd.to_numeric(df["Idade"], errors="coerce")
```

```python
df['Idade']
```

```python
# O comando .loc é uma das propriedades mais importantes e utilizadas do Pandas. Ele serve para localizar e selecionar (ou alterar) dados dentro de um DataFrame baseando-se nos rótulos (labels) das linhas e das colunas.

# O .loc localiza onde a Idade é menor que 0 e altera especificamente a coluna "Idade"

df.loc[df["Idade"] < 0, "Idade"] = np.nan
```

```python
df["Idade"]
```

```python
# Agora consigo fazer cálculos numéricos
media_idades = df["Idade"].mean()
print(f"A média de idade dos usuários é: {media_idades:.1f} anos")
```

```python
import plotly.express as px

fig = px.bar(
    df,
    x="Nome",
    y="Minutos_Assistidos",
    title="Minutos Assistidos por Usuário",
    labels={"Minutos_Assistidos": "Tempo (Minutos)", "Nome": "Usuários"},
    text_auto=True # Mostra o valor exato no topo de cada barra
)

fig.show()
```
