### Funções e métodos

Como as tuplas são imutáveis (não permitem adicionar, remover ou alterar itens), elas têm muito menos métodos do que as listas. Na verdade, elas possuem apenas dois métodos oficiais.

Método count

```python
# Uma tupla com respostas de um forms
respostas = ("Sim", "Não", "Sim", "Sim", "Não")

# Contando quantos "Sim" existem
print(respostas.count("Sim"))
```

Método index

```python
# Uma tupla com os colocados de uma corrida
podio = ("Ana", "Bia", "Carlos", "Lucas")

# Em qual posição a "Bia" aparece?
posicao = podio.index("Bia")

print(f"A Bia apareceu primeiro na posição: {posicao}")
# Saída: 1 (Lembrando que o Python começa a contar do 0)
```

Funções

```python
# Notas de um aluno em uma tupla
notas = (8, 9, 7)

# O sum() soma todos os números dentro da tupla
total_notas = sum(notas)

print(f"A soma das notas é: {total_notas}")
# Saída: 24
```

```python
temperaturas = (15, 28, 22, 19)

maior = max(temperaturas)
menor = min(temperaturas)

print(f"Temperatura máxima: {maior}°C") # Saída: 28
print(f"Temperatura mínima: {menor}°C") # Saída: 15
```

```python
convidados = ("Alice", "Bruno", "Caio", "Duda")

quantidade = len(convidados)

print(f"Existem {quantidade} pessoas na tupla.")
# Saída: 4
```

### Exercício: Software de edição
Um software de edição de imagem recebeu a cor (300, 50, -10). Crie uma função que verifique se os valores da tupla RGB são válidos (entre 0 e 255). Se algum valor estiver fora, avise o usuário.

```python
cor_errada = (300, 50, -10)

for valor in cor_errada:
    if valor < 0 or valor > 255:
        print(f"Erro! O valor {valor} é inválido para RGB.")
```

### Exercício: Processamento de Rota de Entrega

Cenário:
Você tem uma lista de tuplas, onde cada tupla representa um ponto de entrega (Latitude, Longitude). Seu objetivo é:

Contar quantas entregas existem.

Verificar se uma coordenada específica (o Centro de Distribuição) está na lista.

Exibir cada coordenada de forma legível para o motorista.

```python
# Lista de tuplas (cada tupla é um ponto fixo x, y)
rota_entregas = [
    (-23.5505, -46.6333),
    (-23.5589, -46.6611),
    (-23.5505, -46.6333), # Entrega duplicada no mesmo prédio
    (-23.5921, -46.6723)
]
```

```python
# 1. Usando a função len() para saber o total de paradas
total_paradas = len(rota_entregas)
print(f"Total de paradas na rota: {total_paradas}")
```

```python
# 2. Usando o método .count() para ver quantas entregas são no mesmo local
ponto_especifico = (-23.5505, -46.6333)
repeticoes = rota_entregas.count(ponto_especifico)
print(f"Entregas para o ponto {ponto_especifico}: {repeticoes}")
```

```python
# 3. FOR
for destino in rota_entregas:
    print(f"Destino: {destino}")
```

```python
# Adicionando uma nova coordenada (uma nova tupla)
# Note os parênteses extras: append( (x, y) )
rota_entregas.append((-23.6000, -46.7000))
```

```python
print(rota_entregas)
```

### Exercícios complementares

Ex1: Crie uma tupla com as opções de bebidas de uma lanchonete: ("Suco", "Água", "Refrigerante"). Peça para o usuário digitar uma bebida e diga se a lanchonete possui essa opção ou não.

```python
cardapio = ("Suco", "Água", "Refrigerante")

pedido = input("O que deseja beber? ")

# Usando o 'in' para verificar a existência
if pedido in cardapio:
    print("Temos essa bebida no estoque!")
else:
    print("Desculpe, não temos essa opção.")
```

Ex2: Um jogador fez três partidas e você guardou as pontuações na tupla pontos = (120, 150, 90). Use comandos simples para mostrar a maior pontuação e a soma total de pontos.

```python
pontos = (120, 150, 90)

# Usando comandos diretos do Python
print("Sua maior pontuação foi:", max(pontos))
print("Total de pontos acumulados:", sum(pontos))
```
