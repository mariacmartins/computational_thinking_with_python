### Exercício: Software de edição
Um software de edição de imagem recebeu a cor (300, 50, -10). Crie uma função que verifique se os valores da tupla RGB são válidos (entre 0 e 255). Se algum valor estiver fora, avise o usuário.

```python
cor_errada = (300, 50, -10)

for valor in cor_errada:
    if valor < 0 or valor > 255:
        print(f"Erro! O valor {valor} é inválido para RGB.")
```

### Exercício: Processamento de Rota de Entrega

Cenário:
Você tem uma lista de tuplas, onde cada tupla representa um ponto de entrega (Latitude, Longitude). Seu objetivo é:

Contar quantas entregas existem.

Verificar se uma coordenada específica (o Centro de Distribuição) está na lista.

Exibir cada coordenada de forma legível para o motorista.

```python
# Lista de tuplas (cada tupla é um ponto fixo x, y)
rota_entregas = [
    (-23.5505, -46.6333),
    (-23.5589, -46.6611),
    (-23.5505, -46.6333), # Entrega duplicada no mesmo prédio
    (-23.5921, -46.6723)
]
```

```python
# 1. Usando a função len() para saber o total de paradas
total_paradas = len(rota_entregas)
print(f"Total de paradas na rota: {total_paradas}")
```

```python
# 2. Usando o método .count() para ver quantas entregas são no mesmo local
ponto_especifico = (-23.5505, -46.6333)
repeticoes = rota_entregas.count(ponto_especifico)
print(f"Entregas para o ponto {ponto_especifico}: {repeticoes}")
```

```python
# 3. FOR
for destino in rota_entregas:
    print(f"Destino: {destino}")
```
