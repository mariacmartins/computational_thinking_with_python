```python
lista_linguagens = ["Java", "R", "Python", "Javascript"]
#Pergunta para a turma: Como acessar 'Python' dentro da lista de strings criada?
print(lista_linguagens[2])
```

```python
#Como acessar 'Python' e 'Javascript' dentro da lista de strings criada?
#Se você quer extrair os dois itens juntos em uma nova lista, você usa o operador : (dois pontos).
#A regra é: [início : fim], onde o "fim" é exclusivo (ele para um antes).
print(lista_linguagens[2:4])
```

```python
# Também poderia usar
print(lista_linguagens[2:])
```

Índice do item e o ponto de parada do fatiamento

1. Acesso Direto (Índice Único)
Para pegar um único item, o número do índice é exato:

lista[0] -> "Java"

lista[1] -> "R"

lista[2] -> "Python"

lista[3] -> "Javascript" (Funciona perfeitamente!)

2. Para um "pedaço" da lista: O último número deve ser o índice do que você quer + 1.

```python
#E se quisermos acessar todos elementos e ignorar Javascript?
print(lista_linguagens[:3])
```

Por que o Python faz assim?
Existe uma razão matemática elegante para o índice final ser exclusivo. Subtraindo o número final pelo inicial, você tem exatamente a quantidade de itens que vai receber:$$4 (fim) - 2 (início) = 2 \text{ itens na resposta}$$

Se o Python incluísse o 4, você receberia 3 itens, o que bagunçaria a lógica matemática de "tamanho do pedaço".

## Funções e métodos em listas

### Funções:

```python
valores = [10, 5, 20, 15]

# Funções que podemos aplicar em listas
print(f"Quantidade de itens: {len(valores)}")
print(f"Soma total: {sum(valores)}")
print(f"Maior valor: {max(valores)}")
print(f"Menor valor: {min(valores)}")
```

```python
# Criando uma nova lista ordenada sem mexer na original
valores_ordenados = sorted(valores)
print(f"Original: {valores}")
print(f"Nova Ordenada: {valores_ordenados}")
```

```python
# Ordenando em ordem decrescente
valores_ordenados_desc = sorted(valores, reverse=True)
print(valores_ordenados_desc)
```

```python
# Criando uma lista de 0 a 9
numeros = list(range(10))

print(numeros)
```

### Métodos

Adicionando elementos a listas

```python
filmes = ["Matrix", "Interestelar"]

# .append(): Adiciona UM item ao final
filmes.append("O Jogo da Imitação")
print(filmes)
```

```python
# Se eu não quiser sobrescrever a lista original, possuo usar o método .copy()
filmes = ["Matrix", "Interestelar"]

# 1. Cria uma cópia real
lista_nova = filmes.copy()

# 2. Adiciona apenas na cópia
lista_nova.append("O Jogo da Imitação")

print(f"Original: {filmes}")     # ["Matrix", "Interestelar"]
print(f"Nova: {lista_nova}")     # ["Matrix", "Interestelar", "O Jogo da Imitação"]
```

```python
# .insert(): Adiciona em uma posição específica
# ex: (Índice 1)
filmes.insert(1, "Inception")
print(filmes)
```

Removendo elementos

```python
print(filmes)
```

```python
# .remove(): Remove pelo NOME do item (o primeiro que encontrar)
filmes.remove("Inception")
print(filmes)
```

```python
# .pop(): Remove pelo ÍNDICE
filmes.pop(0) # Remove o "Matrix" (índice 0)
print(filmes)
```

Organização e ordem

```python
idades = [10, 50, 2, 60, 25]

# .sort(): Coloca em ordem -> alfabética (A-Z) ou numérica crescente
idades.sort()
print(idades)
```

```python
#.sort(reverse=True): Coloca em ordem decrescente
idades.sort(reverse=True)
print(idades)
```

### Exercícios

**Exercício 1: Inventário de RPG**

Cenário: Seu personagem tem uma mochila: inventario = ["Espada", "Escudo", "Poção"].

- Você encontrou um "Mapa" e uma "Tocha". Adicione-os ao final da lista.

- Um ladrão roubou seu primeiro item! Remova-o.

- Você achou uma "Capa de Invisibilidade" e quer guardá-la na frente de tudo (índice 0).

- Organize sua mochila em ordem alfabética para achar as coisas mais rápido.

```python
inventario = ["Espada", "Escudo", "Poção"]

# 1. Adicionando múltiplos itens
# também daria para usar inventario.extend(["Mapa", "Tocha"])
inventario.append('Mapa')
inventario.append('Tocha')

# 2. Removendo o primeiro item
inventario.pop(0)

# 3. Inserindo no início
inventario.insert(0, "Capa de Invisibilidade")

# 4. Ordenação In-Place
inventario.sort()
print(f"Inventário organizado: {inventario}")
```

**Sistema de Ranking de E-sports**

Desafio: 10min

Cenário: Você tem as pontuações de um torneio:
pontos = [450, 890, 120, 1050, 740].

Quantos jogadores participaram?

O sistema precisa criar um "Top 3". Crie uma nova lista chamada podio com os 3 maiores valores, ordenados do maior para o menor.

Verifique se alguém fez exatamente 500 pontos.

```python
pontos = [450, 890, 120, 1050, 740]
```

```python
# 1. Quantidade
print(f"Total de jogadores: {len(pontos)}")
```

```python
# 2. Criando o pódio (Top 3)
podio = sorted(pontos, reverse=True)
podio = podio[:3]
print(podio)
```

```python
# 3. Contagem
print(f"Jogadores com 500 pontos: {pontos.count(500)}")
```

**Ex2: Fluxo de Checkout**

Cenário: Você está desenvolvendo o carrinho de compras de um site.

A lista carrinho = ["Notebook", "Mouse", "Teclado"].

O cliente decidiu adicionar um "Monitor".

Ele percebeu que o "Mouse" era o modelo errado. Acesse o índice dele e substitua por "Mouse Gamer".

Por fim, o cliente desistiu do primeiro item ("Notebook"). Remova-o da lista.

```python
# Resolução Ex1: Tratamento de dados em sequência de DNA
dna = ["A", "T", "X", "G", "C", "X", "A"]

# 1. Localizando o índice
print(f"O primeiro erro está no índice: {dna.index('X')}")

# 2. Removendo o primeiro 'X'
dna.remove("X")
print(dna)
```

```python
# Resolução Ex2: Fluxo de Checkout
carrinho = ["Notebook", "Mouse", "Teclado"]

# 1. Adicionando múltiplos itens (extend)
carrinho.append('Monitor')

# 2. Substituindo via índice (Mouse está no índice 1)
carrinho[1] = "Mouse Gamer"

# 3. Removendo o primeiro item (índice 0)
carrinho.pop(0)

print(f"Carrinho finalizado: {carrinho}")
```
