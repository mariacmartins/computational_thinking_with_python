# Dicionários

O dicionário é criado usando chaves {} e pares de chave: valor.

As chaves do dicionário precisam ser únicas.
Pense no dicionário como uma lista de contatos no celular: você não pode ter dois contatos com o nome exatamente igual "Mãe", porque se você pedir para o assistente de voz "Ligar para Mãe", o sistema não saberia para qual delas ligar.

```python
# Dicionário de um produto
produto = {
    "nome": "Teclado",
    "preco": 150.00,
    "estoque": 20
}
```

```python
type(produto)
```

Acessando elementos

```python
# Nome do dicionário [Chaves]
print(produto["nome"])
print(produto["preco"])
```

```python
print(produto['nome'])
```

Inserindo e atualizando elementos

```python
# Atribuição simples
# Se a chave não existe, ela é criada. Se existe, o valor é atualizado.
produto["cor"] = "Cinza"     # Insere nova chave
produto["preco"] = 135.00    # Atualiza chave existente
```

```python
print(produto)
```

```python
# Usando o método update
produto.update({"marca": "Logi", "garantia": "12 meses"})
print(produto)
```

Removendo elementos

```python
usuario = {
    "id": 101,
    "nome": "Carlos",
    "email": "carlos@email.com",
    "status": "ativo",
    "admin": False
}
```

```python
# Usando o método pop()
email_removido = usuario.pop("email")

print(f"O email {email_removido} foi excluído da base.")
print(usuario)
```

O pop() do dicionário exclui a chave inteira. Ele remove o "par" (o gancho e o que está pendurado nele). Não tem como usar o pop() do dicionário para deixar a chave lá "vazia".

```python
# Usando o comando del
# Remove o status permanentemente
del usuario["status"]

print(usuario)
```

```python
# Método clear
usuario.clear()

print(usuario)
```

Para remover ou limpar um valor sem excluir a chave, você pode usar a atribuição ou métodos da estrutura interna.

```python
#Se você quer manter a chave lá, mas quer que ela não tenha mais um valor útil, você simplesmente atribui None, 0 ou uma string vazia "".
usuario = {"nome": "Carlos", "pontos": 1500}

# O Carlos perdeu os pontos, mas a chave 'pontos' continua existindo
usuario["pontos"] = 0

# Ou usamos None para indicar que o valor é nulo
usuario["nome"] = None

print(usuario)
```

```python
usuario = {"nome": "Carlos", "pontos": 1500}
usuario['nome'] = None
print(usuario)
```

Navegando no Dicionário (keys, values, items)

Para extrair informações em massa, usamos estes três métodos principais:

keys(): Pega apenas os nomes das chaves.

values(): Pega apenas os dados (valores).

items(): Pega os pares completos.

```python
produto = {
    "nome": "Teclado",
    "preco": 150.00,
    "estoque": 20
}
```

```python
print(produto.keys())
```

```python
print(produto.values())
```

```python
print(produto.items())
```

```python
# Lendo apenas as chaves
for chave in produto.keys():
    print(chave)
```

```python
# Lendo apenas os valores
for valor in produto.values():
    print(valor)
```

```python
# Lendo chave e valor ao mesmo tempo
for chave, valor in produto.items():
    print(f"{chave}: {valor}")
```

### Desafio 2: Gestão de uma Biblioteca Pessoal

(10min)

Você decidiu organizar sua estante de livros usando um dicionário onde a chave é o título do livro e o valor é a quantidade de páginas que você já leu.

meus_livros = {
    "Dom Casmurro": 256,
    "O Hobbit": 310,
    "Pequeno Príncipe": 96,
    "Sherlock Holmes": 120
}

Escreva um código em Python que:

Use um laço for para avaliar os livros e mostrar uma mensagem especial apenas para livros com mais de 200 páginas, dizendo que eles são "Livros Longos" e a mensagem "Livros Curtos" para livros com < 200 páginas.

Output da mensagem:
- {Nome do livro}: {paginas} páginas [Livro Longo]
OU
- {Nome do livro}: {paginas} páginas [Livro Curto]

```python
meus_livros = {
    "Dom Casmurro": 256,
    "O Hobbit": 310,
    "Pequeno Príncipe": 96,
    "Sherlock Holmes": 120
}

# Percorrendo o dicionário
for livro, paginas in meus_livros.items():
    # Exibindo as informações
    if paginas > 200:
        print(f"{livro}: {paginas} páginas [Livro Longo]")
    else:
        print(f"{livro}: {paginas} páginas [Livro Curto]")
```

### Dicionário com Listas (Uma chave, vários valores)
Ideal para quando uma categoria possui vários itens.

```python
# Exemplo de uso de dicionários com listas
carrinho = {
    "cliente": "Maria",
    "itens": ["Mouse", "Teclado", "Monitor"]
}
```

```python
print(carrinho['itens'])
```

```python
print(carrinho["itens"][0]) # Acessa o primeiro item da lista interna, por isso nesse caso em específico usamos o índice
```

```python
carrinho.update({'cliente_clube': True})
print(carrinho)
```

```python
# Podemos aplicar métodos de listas em listas dentro de dict
carrinho["itens"].append("Notebook")
# O dicionário continua o mesmo, a lista interna cresceu.
```

```python
print(carrinho)
```

```python
# Podemos aplicar métodos de listas em listas dentro de dict Remove apenas o elemento da lista, a chave 'itens' permanece lá.
carrinho["itens"].remove("Mouse")
print(carrinho)
```

### Exercício de consolidação de conceitos e entendendo como o for funciona nesse tipo de estrutura

Enunciado: O Painel de Pedidos
Uma cafeteria armazena os pedidos de uma mesa em um dicionário. A chave "cliente" contém o nome da pessoa e a chave "pedido" contém uma lista com os nomes dos cafés e doces escolhidos.

Escreva um código em Python que:

Imprima uma saudação com o nome do cliente.

Use o laço for para percorrer apenas a lista de itens do pedido e mostre cada um deles na tela, um abaixo do outro.

Ao final, mostre quantos itens o cliente pediu no total (dica: use len() na lista).

```python
comanda = {
    "cliente": "Gabriel",
    "pedido": ["Espresso", "Pão de Queijo", "Fatia de Bolo", "Capuccino"]
}

print(f"Olá, {comanda['cliente']}! Seu pedido está sendo preparado:")
print("-" * 30)

# O laço for foca apenas no valor da chave 'pedido'
for item in comanda["pedido"]:
    print(f"-> {item}")

# Calculando o total de itens na lista interna
total = len(comanda["pedido"])

print(f"Total de itens selecionados: {total}")
```
