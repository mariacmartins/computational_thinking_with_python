```python
def busca_linear(arr, alvo):
    # Percorre cada elemento e seu índice na lista
    for i in range(len(arr)):
        if arr[i] == alvo:
            return i  # Retorna o índice onde o elemento foi encontrado

    return -1  # Retorna -1 se o elemento não estiver presente na lista

# Exemplo de uso:
numeros = [36, 12, 45, 9, 22, 88, 15]
alvo = 22

resultado = busca_linear(numeros, alvo)

if resultado != -1:
    print(f"Elemento {alvo} encontrado no índice: {resultado}")
else:
    print(f"Elemento {alvo} não encontrado.")
```

```python
def busca_binaria(lista, item):
    inicio = 0
    fim = len(lista) - 1
    while inicio <= fim:
        meio = (inicio + fim) // 2
        if lista[meio] == item:
            return meio
        elif item > lista[meio]:
            inicio = meio + 1
        else:
            fim = meio - 1
    return -1


# Exemplo de uso:
numeros = [36, 12, 45, 9, 22, 88, 15]

# 1. Ordenar a lista primeiro!
numeros.sort()  # Fica: [9, 12, 15, 22, 36, 45, 88]
alvo = 22

numeros
```

```python
resultado = busca_binaria(numeros, alvo)

if resultado != -1:
    print(f"Elemento {alvo} encontrado no índice: {resultado}")
else:
    print(f"Elemento {alvo} não encontrado.")
```

```python
from datetime import datetime
lista_numeros = list(range(10000000))

tempo_inicial = datetime.now()
print(busca_binaria(lista_numeros, 10000))
tempo_final = datetime.now()
print(f'Tempo Busca Binária: {tempo_final - tempo_inicial}')

tempo_inicial = datetime.now()
print(busca_linear(lista_numeros, 10000))
tempo_final = datetime.now()
print(f'Tempo Busca Linear: {tempo_final - tempo_inicial}')
```
