## Estruturas de repetição - For

```python
# Exemplo 1: Range simples
for i in range(5):
    print(f"Index atual: {i}")
```

```python
# Processando apenas índices pares (de 0 a 10)
for i in range(0, 11, 2):
    print(f"Processando registro par: {i}")
```

```python
# Contagem regressiva
for i in range(10, 0, -1):
    print(f"Timeout em: {i} segundos")
```

```python
# Objetivo: Calcular a média de 4 notas de um aluno

# Imagine que um professor tem 4 notas de um aluno e precisa calcular a média. O for vai servir para pedir cada nota, uma por uma.

soma_notas = 0
total_de_provas = 4

# Queremos que apareça: "Digite a 1ª nota", "Digite a 2ª nota"... até a 4ª.
# Por isso o range vai de 1 até 5 (para incluir o 4)
for i in range(1, 5):
    # O 'i' assume os valores: 1, 2, 3, 4
    nota = float(input(f"Digite a {i}ª nota do aluno: "))

    # Acumulando o valor na variável soma_notas
    soma_notas = soma_notas + nota

# Cálculo final após sair do loop
media = soma_notas / total_de_provas

print(f"--- Resultado Final ---")
print(f"Média do aluno: {media}")
```

```python
# Sistema de "Cálculo de Desconto"
num_products = 4
discount_rate = 0.15

for i in range(1, num_products + 1):
    price = float(input(f"Digite o preço do produto {i}: R$ "))

    # Lógica de negócio: se for caro, aplica desconto
    if price > 100:
        final_price = price * (1 - discount_rate)
        print(f"Desconto aplicado! Novo preço: R$ {final_price:.2f}")
    else:
        print(f"Sem desconto. Preço mantido: R$ {price:.2f}")
```

```python
#LOGIN NO SISTEMA
target_user = "admin_sistema"
max_attempts = 3
is_authenticated = False  # Flag de estado: começa como Falso

print("--- AUTENTICAÇÃO DO SISTEMA ---")

# O loop controla as tentativas de 1 até 3
for attempt in range(1, max_attempts + 1):
    print(f"\nTentativa {attempt} de {max_attempts}")
    input_user = input("Digite seu nome de usuário: ")

    if input_user == target_user:
        print(f"Usuário '{input_user}' reconhecido. Acesso liberado!")
        is_authenticated = True  # Mudamos o estado para Verdadeiro
        break  # Interrompe o loop pois já encontrou o usuário
    else:
        print("Usuário não encontrado no banco de dados.")

# Verificação final: se a flag continuar False, bloqueamos
if not is_authenticated:
    print("SISTEMA BLOQUEADO: Muitas tentativas sem sucesso.")
```

## Percorrendo listas
Veremos listas apenas na próxima aula, mas o for também é muito utilizado pra percorrer listas

```python
server_list = ["srv-01", "srv-02", "srv-03"]

for server in server_list:
    # A variável 'server' assume o valor de cada item por vez
    print(f"Conectando ao servidor: {server}")
```
