### Desafio Prático: Calculador de Frete Inteligente
(15 min)

Contexto:
Sua loja virtual oferece frete grátis para compras acima de um certo valor. Para compras menores, é cobrada uma taxa fixa. Você precisa criar uma função que receba o valor da compra e decida qual o valor final que o cliente deve pagar.

O que deve ser feito:

Crie uma função chamada calculate_shipping.

Essa função deve receber um parâmetro: purchase_amount.

Dentro da função:

Se o valor da compra for maior ou igual a 200 reais, o frete é R$ 0,00.

Se for menor que 200 reais, o frete é uma taxa fixa de R$ 20,00.

A função deve retornar (return) apenas o valor do frete.

No programa principal:

- Peça ao usuário o valor total do carrinho de compras (via input).

- Chame a função para descobrir o valor do frete.

- Calcule o total final (valor da compra + frete).

Exiba os detalhes para o usuário:
print(f"\nResumo do Pedido:")
print(f"Compra: {}")
print(f"Frete: {}")
print(f"Total: {}")

```python
def calculate_shipping(purchase_amount):
    if purchase_amount >= 200.00:
        return 0.00  # Frete Grátis
    else:
        return 20.00 # Taxa Fixa

cart_total = float(input("Digite o valor total da compra: R$ "))

# 1. Descobrimos o frete usando a função
shipping_cost = calculate_shipping(cart_total)

# 2. Calculamos o valor final
final_total = cart_total + shipping_cost

# 3. Exibimos o resumo
print(f"\nResumo do Pedido:")
print(f"Compra: R$ {cart_total:.2f}")
print(f"Frete:  R$ {shipping_cost:.2f}")
print(f"Total:  R$ {final_total:.2f}")
```

### Desafio Prático: Currency Converter

Você está desenvolvendo o módulo financeiro de um sistema internacional. Todos os valores no seu banco de dados devem ser armazenados em Dólar (USD). O usuário pode inserir valores em Real (BRL) ou Euro (EUR). Sua missão é criar uma função que trate o texto digitado pelo usuário, converta para o câmbio correto e retorne o valor pronto para ser salvo.

**O que a função (convert_to_usd) deve fazer:**

Parâmetros: amount (valor) e currency_type (a moeda de origem).

- Substituir vírgulas por pontos (para que o Python entenda como número decimal).

- Remover qualquer caractere que não seja número ou ponto.

- Converter o texto limpo para o tipo decimal (float).

```python
# Usando funções simples aprendidas até agora

def convert_to_usd(amount, currency_type):
    clean_amount = amount.replace(",", ".")

    # Converte a string tratada para número decimal
    value = float(clean_amount)

    if currency_type.upper() == "BRL":
        usd_value = value / 5.00  # Taxa: 1 USD = 5 BRL
    elif currency_type.upper() == "EUR":
        usd_value = value * 1.10  # Taxa: 1 EUR = 1.10 USD
    else:
        usd_value = value

    return usd_value

# --- PROGRAMA PRINCIPAL ---
print("--- CURRENCY CONVERTER TOOL ---")
user_value = input("Coloque o valor a ser convertido: ")
user_currency = input("Moeda (EUR/BRL): ")

final_price_usd = convert_to_usd(user_value, user_currency)

# Exibindo o valor que seria enviado ao Banco de Dados
print(f"\n Valor final a ser salvo no banco de dados: ${final_price_usd}")
```

Mas e se fosse digitado R$ 5 no input? Esse código daria um erro!

Nesse sentido, podemos usar a biblioteca re para apoiar nesse processo.

O que é a biblioteca re?
A biblioteca re (Regular Expressions, ou Expressões Regulares) é uma ferramenta poderosa para buscar, extrair ou modificar padrões de texto.

Uma Expressão Regular é uma sequência de caracteres que forma um padrão de busca. É como se você desse um "scanner" para o computador e dissesse: "Encontre tudo o que se pareça com isso".

Artigo para ler: https://medium.com/pyladiesbh/regex-b%C3%A1sico-em-python-31dcb7fac046

```python
import re

def convert_to_usd(amount, currency_type):
    # Mantém apenas números, pontos e vírgulas. Remove símbolos (R$, €), letras e espaços extras.
    # O nome vem de "Substitute" (Substituir). Ela funciona como um "Localizar e Substituir"
    # O Padrão (r'[^0-9,.]'): É a regra de busca. []: Define uma lista de permitidos. ^: Diz "Tudo que NÃO for isso". 0-9,.: São os números, a vírgula e o ponto.

    clean_amount = re.sub(r'[^0-9,.]', '', amount)

    # 2. Padronização: Troca vírgula por ponto para o float() funcionar
    clean_amount = clean_amount.replace(",", ".")

    # 3. Conversão para decimal
    value = float(clean_amount)

    # 4. Lógica de Câmbio
    currency = currency_type.upper().strip()

    if currency == "BRL":
        usd_value = value / 5.00
    elif currency == "EUR":
        usd_value = value * 1.10
    else:
        usd_value = value

    return usd_value

# --- PROGRAMA PRINCIPAL ---
print("--- CURRENCY CONVERTER TOOL ---")
user_value = input("Coloque o valor a ser convertido: ")
user_currency = input("Moeda (EUR/BRL): ")

final_price_usd = convert_to_usd(user_value, user_currency)

# Exibindo o valor formatado
print(f"\n Valor final a ser salvo no banco de dados: ${final_price_usd}")
```

## Variáveis locais e globais

1. Variável Global:
- Fora de qualquer função (no corpo principal do código).
- O programa inteiro consegue enxergá-la.
- Serve para dados que várias partes do código precisam acessar.

2. Variável Local:
- Dentro do bloco de uma função.
- Ela é perdida assim que a função termina a execução.
- Uso: Cálculos temporários e auxiliares que não interessam ao resto do programa.

Regra LEGB
Quando você chama o nome de uma variável dentro de uma função, o Python faz uma busca em "camadas" para decidir qual valor usar. Ele usa a regra LEGB (do nível mais interno para o mais externo):

L (Local): Primeiro, ele procura se a variável foi criada dentro da função atual.

E (Enclosing): Se não achou, ele olha se ela existe em uma função "pai" (no caso de funções uma dentro da outra).

G (Global): Se ainda não achou, ele procura no nível principal do arquivo (fora de todas as funções).

B (Built-in): Por último, ele checa se é uma palavra reservada do próprio Python (como print, sum, len).


```python
## Entendendo variáveis locais e globais
def calcular_imposto(preco):
    taxa = 0.05  # Variável LOCAL
    return preco * taxa

print(calcular_imposto(100))
```

```python
# Erro! Se tentarmos usar 'taxa' aqui fora, o Python dirá que ela não existe. Pois é um variável que só existe dentro da função
print(taxa)
```

```python
def func1():
  x = 1
  print(x)

def func2():
  x = 2
  print(x)
```

```python
func1()
func2()
```

```python
# Variável global (fora da função)
user = "Clara"

def boas_vindas(): # Sem parâmetro
    print(f"Olá, {user}!") # Lendo direto da global

boas_vindas() # Sem passar nada
```

Lógica:

"Existe uma variável chamada user criada dentro desta função?"
Resposta: Não.

"Existe uma variável chamada user no corpo principal do arquivo (fora das funções)?"

Resposta: Sim! Ela vale "Clara".

Então, o Python "pega emprestado" o valor da global porque não encontrou uma local com o mesmo nome.

```python
usuario = "Clara"

def boas_vindas(name): # Usando parâmetro
    print(f"Olá, {name}!")

boas_vindas(usuario)
```

E se os nomes fossem iguais?

Se você criar uma variável dentro da função com o mesmo nome da global, o Python dará prioridade para a Local.

```python
user = "Clara"  # Global

def boas_vindas():
    user = "Betina"  # Local (Tem o mesmo nome, mas é outra variável)
    print(f"Olá, {user}!")

boas_vindas()  # Vai imprimir: Olá, Betina!
print(user)    # Vai imprimir: Clara (A global continua intacta)
```

## Parâmetros posicionais, nomeados e default

Posicionais: Como o nome diz, a ordem importa. O Python entende qual valor pertence a qual variável baseando-se apenas na posição (quem vem primeiro, segundo, etc.).

Nomeados: Aqui, você diz explicitamente o nome do parâmetro seguido do valor (chave=valor). A ordem deixa de importar, pois o Python usa o nome para "ligar os pontos".

Default: Muitas vezes, queremos que um parâmetro seja opcional. Nós definimos um valor "de reserva" na criação da função.

```python
# Parâmetros posicionais
def criar_perfil(nome, idade, profissao):
    print(f"Nome: {nome} | Idade: {idade} | Profissão: {profissao}")

# Aqui a ordem é OBRIGATÓRIA: Nome, Idade, Profissão
criar_perfil("Maria Clara", 29, "Manager")
```

Problema: Se você trocar a ordem na função criar_perfil, o Python vai aceitar, mas os dados estarão errados.

**E quando não sabemos quantos parâmetros precisamos receber ou quais são eles?**

O  "* args"  é um parâmetro de empacotamento. O nome vem de arguments (argumentos), e o asterisco (*) é o que faz a mágica.

Ele permite que uma função receba um número indefinido de argumentos posicionais. Quando você usa o *args, o Python pega todos os valores extras que foram passados na chamada da função e os coloca dentro de uma Tupla.

Parâmetros Fixos: São como cadeiras numeradas. Se você tem 2 cadeiras (a, b), só podem sentar 2 pessoas. Se vierem 3, o código quebra.

Parâmetros Default: São cadeiras reservadas. Se a pessoa não vier, a cadeira já tem um valor padrão nela.

*args: É um sofá infinito. Não importa quantas pessoas cheguem, todas sentam juntas no mesmo sofá (a Tupla).

Quando você coloca o * no parâmetro da função, você está dizendo ao Python para abrir uma mala e recolher tudo o que o usuário jogar.

```python
def somar_vendas(*args):
    print(args)
    total = sum(args)
    print(f"Total de vendas: R$ {total}")

somar_vendas(10, 50)          # Funciona com 2
somar_vendas(10, 20, 30, 40) # Funciona com 4
```

```python
# Dará um erro sem o *
def somar_vendas(args):
    total = sum(args)
    print(f"Total de vendas: R$ {total}")

somar_vendas(10, 50)          # Funciona com 2
somar_vendas(10, 20, 30, 40) # Funciona com 4
```

## Anotações de tipo (type hints)

As anotações de tipo (ou Type Hints) foram introduzidas no Python 3.5 e mudaram a forma como escrevemos código profissional. Embora o Python continue sendo uma linguagem de tipagem dinâmica (você não é obrigado a declarar tipos), as anotações servem como um "contrato" que deixa suas intenções claras.

Imagine que você recebe uma função chamada processar_dados(x). O que é x? Um número? Uma lista? Um texto? Sem anotações, você precisa ler todo o código ou a documentação para descobrir.

As anotações servem para:

Documentação Automática: O código explica a si mesmo.

Melhoria no Editor (IDE): O VS Code ou PyCharm conseguem te dar sugestões (autocompletar) muito mais precisas.

**Python vs. Outras Linguagens**

C++ / Java (Tipagem Estática e Forte): Se você declarar int x = 10; e tentar passar uma string, o código nem compila. O tipo é obrigatório e rígido.

Python com Type Hints: Se você declarar x: int = 10, o Python ainda aceita que você mude para x = "texto" depois.

Conclusão: No Python, as anotações são informativas, não impositivas. Elas não mudam como o Python roda o código, elas apenas ajudam as ferramentas de desenvolvimento e outros programadores.

Ou seja, em Python, essas expressões não possuem significado algum em termos programáticos, e são
ignoradas pelo interpretador da linguagem, assim como ocorre com os comentários
acrescidos ao código.

A sintaxe básica usa : para parâmetros e -> para o retorno da função.

def nome_função(arg1: tipo, arg2: tipo) -> tipo_retorno

Se você precisa que uma função retorne tipos diferentes dependendo da situação (por exemplo: um dicionário quando dá tudo certo ou uma string de erro quando algo falha), usamos o operador de união (|).

```python
def create_user(name: str, age: int, city: str) -> dict | str:
    """
    Tenta formatar os dados de um usuário ou retorna um erro se a idade for inválida.

    Args:
        name (str): O nome completo do usuário.
        age (int): A idade do usuário (deve ser maior que zero).
        city (str): A cidade onde o usuário reside.

    Returns:
        dict | str: Retorna um dicionário com os dados em caso de sucesso
                    ou uma string contendo a mensagem de erro.
    """
    # Validação simples
    if age <= 0:
        return "Erro: A idade deve ser um número positivo."

    user_data = {
        "name": name,
        "age": age,
        "city": city
    }

    return user_data

# --- Testando os dois tipos de retorno ---

# Cenário 1: Retorna um dicionário (dict)
sucesso = create_user("Alice", 25, "Gramado")
print(f"Resultado 1: {sucesso} | Tipo: {type(sucesso)}")

# Cenário 2: Retorna uma mensagem de erro (str)
erro = create_user("Bob", -5, "Natal")
print(f"Resultado 2: {erro} | Tipo: {type(erro)}")
```

try: Onde você coloca o código a ser testado.

except NomeDoErro: Trata um erro específico.

except Exception: Trata qualquer erro (use com moderação).

as e: Serve para você "ler" a mensagem de erro que o Python gerou.

```python
# Exemplo com except específico simples
def realizar_divisao(numero, divisor):
    try:
        resultado = numero / divisor
        print(f"O resultado é: {resultado}")

    except ZeroDivisionError:
        # O que fazer se o erro ocorrer
        print("Erro: Não é possível dividir um número por zero.")

# --- Testes ---

print("--- Teste 1 (Divisão válida) ---")
realizar_divisao(10, 2)

print("\n--- Teste 2 (Divisão por zero) ---")
realizar_divisao(10, 0)
```

```python
# Exemplo com mais de except específico

def acessar_lista(indice: int) -> None:
    cores = ["Verde", "Amarelo", "Azul"]

    try:
        print(f"A cor escolhida foi: {cores[indice]}")
    except IndexError:
        # Erro específico para quando a posição não existe na lista
        print(f"Erro: O índice {indice} está fora da lista de cores.")
    except TypeError:
        # Erro específico para quando o dado passado não é um número
        print("Erro: Você precisa digitar um número inteiro para o índice.")

# Testes
acessar_lista(1)     # Funciona (Amarelo)
acessar_lista(10)    # Erro de índice
acessar_lista("um")  # Erro de tipo
```

```python
# Exemplo com exception genérico

def acessar_lista_generica(indice: int) -> None:
    cores = ["Verde", "Amarelo", "Azul"]

    try:
        # O código tenta ser executado
        print(f"A cor escolhida foi: {cores[indice]}")

    except Exception as e:
        # Qualquer erro (índice fora da lista, tipo errado, etc.) cai aqui
        print(f"Ocorreu um erro ao tentar acessar a lista!")
        print(f"Detalhe técnico do erro: {e}")

# --- Testes ---

print("--- Teste 1 (Sucesso) ---")
acessar_lista_generica(1)

print("\n--- Teste 2 (Índice inexistente) ---")
acessar_lista_generica(10)

print("\n--- Teste 3 (Tipo de dado errado) ---")
# Nota: Como o type hint pede int, o editor pode avisar, mas o Python tentará rodar
acessar_lista_generica("um") # type: ignore
```

No tratamento específico, você prepara uma ação exata para cada tipo de problema.

O tratamento genérico é como dizer: "Se acontecer qualquer problema, apenas me avise que algo deu errado".

```python
# Combinando as 2 abordagens:
def dividir_conta_total(total: str, pessoas: str) -> float | str:
    try:
        valor_total = float(total)
        quantidade_pessoas = int(pessoas)

        return valor_total / quantidade_pessoas

    except ValueError:
        return "Erro: Os valores de entrada devem ser números."

    except ZeroDivisionError:
        return "Erro: A quantidade de pessoas não pode ser zero."

    except Exception as e:
        return f"Erro inesperado: {e}"

# --- Testando e capturando o return ---

# Caso 1: Sucesso
resultado1 = dividir_conta_total("200", "4")
print(f"Resultado 1: {resultado1}") # Imprime 50.0

# Caso 2: Erro Específico
resultado2 = dividir_conta_total("200", "0")
print(f"Resultado 2: {resultado2}") # Imprime a mensagem do ZeroDivisionError

# Caso 3: Erro Genérico (ex: passando algo nulo)
resultado3 = dividir_conta_total(None, "5") # type: ignore
print(f"Resultado 3: {resultado3}") # Imprime a mensagem do Exception
```

```python
# mantendo a abordagem genérica:
def dividir_conta_total(total: str, pessoas: str) -> float | str:
    try:
        valor_total = float(total)
        quantidade_pessoas = int(pessoas)

        return valor_total / quantidade_pessoas

    except Exception as e:
        return f"Erro: {e}"

# --- Testando e capturando o return ---

print(dividir_conta_total("200", "0"))
```

### raise
O raise é uma instrução que força a interrupção do fluxo de execução atual e cria um objeto de exceção.

Para a execução da linha atual imediatamente.

Sobe na "pilha de chamadas" (Call Stack) procurando por um bloco try/except que saiba lidar com aquele tipo de erro.

Se não encontrar nenhum tratamento, o programa encerra e exibe o Traceback (aquele erro detalhado no terminal).

Quando usar?

Validação de Regras de Negócio: Quando o dado é tecnicamente correto (é um número), mas logicamente errado (é uma idade negativa).

Proteção de Funções: Quando você quer garantir que quem usar sua função passe os dados exatamente como você pediu. Se ele não passar, você "pune" o erro com um raise.

```python
def realizar_saque(valor):
    saldo = 100

    try:
        if valor > saldo:
            # O raise 'lança' o erro de propósito
            raise ValueError("Saldo insuficiente!")

        # Esta linha só executa se o raise NÃO for acionado
        print(f"Saque de R${valor} realizado com sucesso!")

    except Exception as e:
        # O except captura a mensagem que o raise lançou
        print(f"Erro no sistema: {e}")

# Testando a função
realizar_saque(150)
```

```python
def realizar_saque(valor):
    saldo = 100

    try:
        if valor > saldo:
            # O raise 'lança' o erro de propósito
            raise ValueError('Saldo insuficiente')

        # Esta linha só executa se o raise NÃO for acionado
        print(f"Saque de R${valor} realizado com sucesso!")

    except Exception as e:
        # O except captura a mensagem que o raise lançou
        print(f"Erro no sistema: {e}, type: {type(e)}")

# Testando a função
realizar_saque(150)
```

No mundo real, tratar erro é só dar print?
Definitivamente não. No mundo real, um print é quase inútil porque ninguém está olhando o console do servidor 24 horas por dia. Se o seu código der um erro às 3 da manhã, o print vai sumir no histórico.

No desenvolvimento profissional, usamos outras estratégias:

Logging: O erro é salvo em um arquivo permanente ou em um sistema de monitoramento (como Datadog). Isso permite que os desenvolvedores vejam o erro, o horário e o que o causou, mesmo dias depois.

Retorno de Status: Em APIs, o erro é retornado como um código HTTP (ex: 400 Bad Request ou 404 Not Found) para que o sistema que fez o pedido saiba como reagir.

Recuperação de Falhas (Retry):  O tratamento pode ser tentar novamente 3 vezes antes de desistir.

Fallback: Se uma função falha ao pegar o preço atualizado de um produto, o tratamento pode ser usar o último preço salvo no banco de dados para não deixar o usuário sem resposta.
