# JSON
Criando um objeto em Python

```python
aluno = {
    "nome": "Maria",
    "idade": 22,
    "curso": "Python"
}

print(aluno)
```

```python
print(type(aluno))
```

Transformando o dict em JSON

```python
import json

aluno = {
    "nome": "Maria",
    "idade": 22,
    "curso": "Python"
}

json_aluno = json.dumps(aluno)

print(json_aluno)
```

```python
print(type(json_aluno))
```

```python
# deixando o json formatado
json_aluno = json.dumps(
    aluno,
    indent=4
)

print(json_aluno)
```

JSON -> PYTHON

```python
import json

texto = '''
{
    "nome": "Carlos",
    "idade": 25,
    "curso": "Python"
}
'''
```

```python
aluno = json.loads(texto)

print(aluno)
print(type(aluno))
```

Manipulando o dict depois de carregar o JSON

```python
print(aluno["nome"])
```

Depois que o JSON é convertido para Python, podemos manipulá-lo normalmente como um dict.

```python
print(type(json_cursos))
```

Lista de objetos

```python
alunos = [
    {
        "nome": "Ana",
        "idade": 20
    },
    {
        "nome": "Carlos",
        "idade": 22
    },
    {
        "nome": "João",
        "idade": 19
    }
]
```

```python
# transformando pra json

json_alunos = json.dumps(
    alunos,
    indent=4,
    ensure_ascii=False
)

print(json_alunos)
```

```python
type(json_alunos)
```

Criando um arquivo JSON

```python
import json

dados = {
    "alunos": [
        {
            "nome": "Ana",
            "idade": 20,
            "curso": "Python"
        },
        {
            "nome": "Carlos",
            "idade": 22,
            "curso": "SQL"
        }
    ]
}

with open("alunos.json", "w", encoding="utf-8") as arquivo:
    json.dump(dados, arquivo, indent=4, ensure_ascii=False)
```

```python
# outra forma:
"""
alunos = [
    {
        "nome": "Ana",
        "idade": 20,
        "curso": "Python"
    },
    {
        "nome": "Carlos",
        "idade": 22,
        "curso": "SQL"
    }
]
"""
```

Lendo um arquivo JSON

```python
# Carregar (deserializar) o arquivo JSON
with open("alunos.json", "r", encoding="utf-8") as arquivo:
    dados = json.load(arquivo)
```

```python
type(dados)
```

```python
dados
```

```python
# O objeto 'dados' é um dicionário contendo a chave "alunos"
lista_alunos = dados["alunos"]
lista_alunos
```

```python
type(lista_alunos)
```

```python
print("--- 1. Listar todos os alunos ---")
for aluno in lista_alunos:
    print(f"Nome: {aluno['nome']} | Curso: {aluno['curso']}")
```

Tratando JSON inválido

```python
texto = '''
{
    "nome": "Maria",
    "idade": 22
'''

try:

    aluno = json.loads(texto)

except Exception:

    print(f"JSON inválido!")
```

```python
texto = '''
{
    "nome": "Maria",
    "idade": 22
'''

try:

    aluno = json.loads(texto)

except json.JSONDecodeError as e:

    print(f"JSON inválido! Erro: {e}")
```

## Exercícios

### Exercício 1 — Criando um JSON
Enunciado

Crie um dict chamado produto com:

id
nome
preco
estoque

Converta o dict para JSON e mostre o resultado com indentação de 4 espaços.

```python
import json

produto = {
    "id": 1,
    "nome": "Notebook",
    "preco": 3500,
    "estoque": 10
}

produto_json = json.dumps(
    produto,
    indent=4
)

print(produto_json)
```

Enunciado

Considere o JSON:

texto = '''
