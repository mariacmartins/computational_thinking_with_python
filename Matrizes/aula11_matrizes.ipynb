# Matrizes

### Criando matrizes

```python
# Criando uma matriz 3x3 (Linhas x Colunas)
matriz = [
    [1, 2, 3], # Linha 0
    [4, 5, 6], # Linha 1
    [7, 8, 9]  # Linha 2
]
```

```python
print(matriz)
```

```python
# Criando matriz com input de usuários

matriz = []

# Loop para as LINHAS (i)
for i in range(3): #qtd de colunas
    linha = [] # Cria uma lista nova para cada andar

    # Loop para as COLUNAS (j)
    for j in range(3): # qtd de colunas
        # Pedimos o valor e mostramos a coordenada para o usuário não se perder
        valor = int(input(f"Digite o valor para a posição [{i}][{j}]: "))
        linha.append(valor) # Adiciona o valor digitado na linha

    matriz.append(linha) # Adiciona a linha completa na matriz

# Exibindo a matriz final de forma organizada
print(matriz)
```

```python
# Criando uma matriz 3x5 com números aleatórios
import random

# preenche a matriz
matriz = []
for i in range(3): #qtd de linhas
    linha = []
    for j in range(5): # qtd de colunas
        n = random.randint(1, 50)
        linha.append(n)
    matriz.append(linha)

print(matriz)
```

**Acessando Elementos**

Para acessar, usamos dois colchetes: matriz[linha][coluna].

Para pegar uma linha inteira: Use apenas um índice.

Para pegar um valor específico: Use os dois.

```python
matriz = [
    [1, 2, 3], # Linha 0
    [4, 5, 6], # Linha 1
    [7, 8, 9]  # Linha 2
]
```

```python
print(matriz[0])
```

```python
print(matriz[1][2])  # Saída: 6 (Linha 1, Coluna 2)
```

**Percorrendo elementos**

```python
matriz = [[1, 2, 3],
[4, 5, 6]]

# Percorrer linhas
for linha in matriz:
  print(linha)
```

```python
# Percorrer itens
for linha in matriz:
  for item in linha:
    print(item)
```

**Alterando Valores**

Basta acessar a "coordenada" e atribuir um novo valor, como em uma variável comum.

```python
print(matriz[1][1])
```

```python
# Mudar o número 5 (centro) para 0
matriz[1][1] = 0
print(matriz[1]) # Saída: [4, 0, 6]
```

Substituindo linhas

```python
# Substituindo a linha 1 (a do meio) por uma lista nova
matriz[1] = [10, 20, 30]
```

```python
print(matriz)
```

Substituindo colunas

```python
matriz = [
    [10, 5, 100], # Linha 0
    [20, 5, 200], # Linha 1
    [30, 2, 300]  # Linha 2
]
```

```python
# Mudando a coluna com valores diferentes
# Aqui, o nosso alvo é a Coluna 1
# len(matriz): Conta quantas linhas a matriz tem. No exemplo, são 3.

novos_valores = [77, 88, 99]

for i in range(len(matriz)):
  matriz[i][1] = novos_valores[i] # O comando central é: matriz[i][1] = novos_valores[i].

print(matriz)
```

1ª Volta (i = 0):
Lado esquerdo: matriz[0][1] → Mira no elemento da Linha 0, Coluna 1

Lado direito: novos_valores[0] → Pega o primeiro valor da lista (77).

2ª Volta (i = 1):
Lado esquerdo: matriz[1][1] → Mira no elemento da Linha 1, Coluna 1

Lado direito: novos_valores[1] → Pega o segundo valor da lista (88).

3ª Volta (i = 2):
Lado esquerdo: matriz[2][1] → Mira no elemento da Linha 2, Coluna 1

Lado direito: novos_valores[2] → Pega o terceiro valor da lista (99).

**Operações em Matrizes**

```python
print(matriz)
```

```python
# Somando o valor 10 a todos os elementos da matriz
# Inicia um laço que percorre cada linha (sublista) dentro da matriz
for linha in matriz:

    # Para a linha atual, cria um índice 'i' que vai de 0 até o tamanho dessa linha
    for i in range(len(linha)):

        # Acessa o elemento na posição 'i' da linha atual e soma 10 ao seu valor
        # Como 'linha' é uma referência à sublista da matriz, a alteração é real
        linha[i] += 10
```

```python
print(matriz)
```

```python
# Como somar o valor 10 a apenas um elemento? Ex: Elemento 13
#O 13 está na primeira linha, então o índice é 0.
#O 13 está na terceira coluna, então o índice é 2.
matriz[0][2] += 10
```

```python
print(matriz)
```
