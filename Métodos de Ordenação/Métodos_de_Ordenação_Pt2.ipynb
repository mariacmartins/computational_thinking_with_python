```python
def merge_sort(arr):
    """
    Função principal do Merge Sort.
    Segue a estratégia 'Dividir para Conquistar': divide a lista recursivamente
    em metades menores até que cada sublista tenha 0 ou 1 elemento.
    """
    # CASO BASE: Se a lista tiver 0 ou 1 elemento, ela já está obrigatoriamente ordenada.
    # Isso interrompe as chamadas recursivas, evitando loops infinitos.
    if len(arr) <= 1:
        return arr

    # -------------------------------------------------------------
    # PASSO 1: DIVIDIR
    # -------------------------------------------------------------
    # Encontra o índice exato do meio da lista usando divisão inteira (//)
    mid = len(arr) // 2

    # Divide a lista em duas metades usando o fatiamento (slicing) do Python
    left_half = arr[:mid]   # Pega do início até o elemento anterior ao meio
    right_half = arr[mid:]  # Pega do meio até o final da lista

    # -------------------------------------------------------------
    # PASSO 2: CONQUISTAR (Chamadas Recursivas)
    # -------------------------------------------------------------
    # Aplica o merge_sort em cada metade de forma independente.
    # O Python vai empilhando essas chamadas até que todas as sublistas
    # se reduzam a tamanhos unitários (atingindo o caso base).
    left_sorted = merge_sort(left_half)
    right_sorted = merge_sort(right_half)

    # -------------------------------------------------------------
    # PASSO 3: COMBINAR (Mesclar)
    # -------------------------------------------------------------
    # Junta as duas metades que já estão ordenadas em uma única lista organizada.
    return merge(left_sorted, right_sorted)


def merge(left, right):
    """
    Função auxiliar responsável por mesclar (fundir) duas listas
    já ordenadas (left e right) em uma nova lista totalmente ordenada.
    """
    sorted_arr = []  # Lista temporária que vai armazenar os elementos em ordem
    i = 0            # Ponteiro para rastrear a posição atual na lista 'left'
    j = 0            # Ponteiro para rastrear a posição atual na lista 'right'

    # Compara os elementos das duas listas um a um enquanto ambas tiverem itens
    while i < len(left) and j < len(right):
        # Se o elemento da esquerda for menor ou igual ao da direita, ele entra primeiro
        if left[i] <= right[j]:
            sorted_arr.append(left[i])
            i += 1  # Avança o ponteiro da lista esquerda
        else:
            # Caso contrário, o elemento da direita é o menor
            sorted_arr.append(right[j])
            j += 1  # Avança o ponteiro da lista direita

    # -------------------------------------------------------------
    # RESTOS DAS LISTAS
    # -------------------------------------------------------------
    # Quando o loop acima termina, uma das listas (left ou right) já foi
    # totalmente consumida, mas a outra ainda pode ter elementos restantes
    # (por serem maiores que todos os outros já inseridos).
    # O comando .extend() joga o restante diretamente no fim da lista ordenada.
    sorted_arr.extend(left[i:])
    sorted_arr.extend(right[j:])

    return sorted_arr


# =============================================================
# Exemplo prático de uso:
# =============================================================
numeros = [36, 12, 45, 9, 22, 12, 88, 15]
ordenados = merge_sort(numeros)

print("Lista original:  [36, 12, 45, 9, 22, 12, 88, 15]")
print("Lista ordenada:", ordenados)
# Saída esperada: [9, 12, 12, 15, 22, 36, 45, 88]
```

```python
def quicksort(arr):
    # Caso base: listas com 0 ou 1 elemento já estão ordenadas
    if len(arr) <= 1:
        return arr

    # Escolhe o pivô como o primeiro elemento da lista
    pivot = arr[0]

    left = []
    middle = []
    right = []

    # Particionamento usando um loop tradicional
    for x in arr:
        if x < pivot:
            left.append(x)
        elif x == pivot:
            middle.append(x)
        else:
            right.append(x)

    # Chamada recursiva e combinação dos resultados
    return quicksort(left) + middle + quicksort(right)

# Exemplo de uso:
numeros = [36, 12, 45, 9, 22, 12, 88, 15]
ordenados = quicksort(numeros)
print("Lista ordenada:", ordenados)
# Saída: [9, 12, 12, 15, 22, 36, 45, 88]
```

```python
import pandas as pd

# Criando um DataFrame de exemplo
dados = {
    'Produto': ['Notebook', 'Mouse', 'Teclado', 'Monitor', 'Cadeira', 'Headset'],
    'Preço': [3500.0, 150.0, 250.0, 1200.0, 900.0, 300.0]
}
df = pd.DataFrame(dados)

print("--- DataFrame Original ---")
print(df)

# -------------------------------------------------------------
# 1. Usando o Quicksort nativo do Pandas (já é o padrão)
# -------------------------------------------------------------
df_quicksort = df.sort_values(by='Preço', kind='quicksort')

print("\n--- Ordenado usando Quicksort Nativo (kind='quicksort') ---")
print(df_quicksort)

# -------------------------------------------------------------
# 2. Usando o Mergesort nativo do Pandas
# -------------------------------------------------------------
# O mergesort é uma escolha útil quando você precisa de uma ordenação ESTÁVEL
# (ou seja, ele preserva a ordem original de elementos que possuem o mesmo valor).
df_mergesort = df.sort_values(by='Preço', kind='mergesort')

print("\n--- Ordenado usando Mergesort Nativo (kind='mergesort') ---")
print(df_mergesort)
```

```python
import time
import pandas as pd
import numpy as np

# Criando um DataFrame grande com 1.0000.000 números aleatórios
df = pd.DataFrame({'Valor': np.random.randint(1, 1000000, size=10000000)})

# 1. Medindo o Pandas com Quicksort
df_q = df.copy()
t0 = time.time()
df_q.sort_values(by='Valor', kind='quicksort', inplace=True)
tempo_quick = (time.time() - t0) * 1000

# 2. Medindo o Pandas com Mergesort
df_m = df.copy()
t0 = time.time()
df_m.sort_values(by='Valor', kind='mergesort', inplace=True)
tempo_merge = (time.time() - t0) * 1000


print(f"Pandas com Quicksort: {tempo_quick:.2f} ms")
print(f"Pandas com Mergesort: {tempo_merge:.2f} ms")
```
