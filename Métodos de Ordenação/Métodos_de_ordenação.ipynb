```python
def bubble_sort(lista):
    """
    Ordena uma lista usando o Bubble Sort.
    """

    # Guarda a quantidade de elementos da lista.
    n = len(lista)

    # i representa a PASSADA que estamos realizando.
    # Na primeira passada, i = 0.
    # Na segunda passada, i = 1.
    #
    # A cada passada, um elemento maior consegue "borbulhar" para o final da parte ainda não ordenada.
    for i in range(n - 1):

        print(f"\n--- PASSADA {i + 1} ---")

        # j representa a posição do elemento que estamos analisando naquele momento.
        #
        # lista[j]     -> elemento atual
        # lista[j + 1] -> elemento vizinho
        for j in range(n - 1 - i):

            print(
                f"j = {j}: comparando {lista[j]} e {lista[j + 1]}")

            # Se o elemento atual for maior que o próximo,
            # significa que eles estão na ordem errada.
            if lista[j] > lista[j + 1]:

                print("  -> Estão fora de ordem. Trocando!")

                # Troca os dois elementos.
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

                print(f"  -> Lista agora: {lista}")

            else:
                print("  -> Já estão na ordem correta.")

        print(f"Resultado da passada {i + 1}: {lista}")

    return lista


numeros = [70,90,3,0,100,2]

print("LISTA ORIGINAL:")
print(numeros)

resultado = bubble_sort(numeros)

print("\nLISTA ORDENADA:")
print(resultado)
```

```python
def selection_sort(lista):
    """
    A cada rodada, procura o menor elemento
    da parte ainda não ordenada e coloca na posição correta.
    """

    n = len(lista)

    # i representa a posição que queremos preencher.
    # Ex.: i=0 → colocar o menor elemento na posição 0.
    for i in range(n - 1):

        # Inicialmente, consideramos que o menor
        # está na posição i.
        menor = i

        # j percorre o restante da lista procurando
        # um elemento menor.
        for j in range(i + 1, n):

            if lista[j] < lista[menor]:
                # Guardamos o ÍNDICE do novo menor.
                menor = j

        # Troca o elemento da posição i
        # pelo menor elemento encontrado.
        aux = lista[i]
        lista[i] = lista[menor]
        lista[menor] = aux

        print(f"i={i}: {lista}")

    return lista


numeros = [70,90,3,0,100,2]

print("Original:", numeros)
print("Ordenada:", selection_sort(numeros))
```

```python
def insertion_sort(lista):
    """
    Pega um elemento por vez e o insere
    na posição correta da parte já ordenada.
    """

    n = len(lista)

    # i representa o elemento que vamos inserir
    # na parte que já está ordenada.
    for i in range(1, n):

        # Guarda o valor que queremos posicionar.
        atual = lista[i]

        # j começa na posição de "atual"
        # e vai andando para a esquerda.
        j = i

        # Enquanto houver elementos à esquerda
        # maiores que "atual", deslocamos esses
        # elementos uma posição para a direita.
        while j > 0 and lista[j - 1] > atual:

            lista[j] = lista[j - 1]

            # Anda uma posição para a esquerda.
            j -= 1

        # Coloca "atual" na posição correta.
        lista[j] = atual

        print(f"i={i}: {lista}")

    return lista


numeros = [70,90,3,0,100,2]

print("Original:", numeros)
print("Ordenada:", insertion_sort(numeros))
```
