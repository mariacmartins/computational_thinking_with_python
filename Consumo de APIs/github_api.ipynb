Doc oficial: https://docs.github.com/en/rest?apiVersion=2026-03-10

```python
import requests
```

```python
# Testando endpoint de users
usuario = "mariacmartins"
url_users = f"https://api.github.com/users/{usuario}"

resposta = requests.get(url_users)

if resposta.status_code == 200:
    dados = resposta.json()
    print(f"Nome: {dados.get('name')}")
    print(f"Bio: {dados.get('bio')}")
    print(f"Repositórios públicos: {dados.get('public_repos')}")
else:
    print("Usuário não encontrado.")
```

```python
print(dados)
```

```python
# Testando endpoint de users
url_repos = f"https://api.github.com/users/{usuario}/repos"

resposta = requests.get(url_repos)
repositorios = resposta.json()

# Exibe os 3 primeiros repositórios e suas linguagens
for repo in repositorios[:3]:
    print(f"- {repo['name']} | Linguagem: {repo['language']}")
```

```python
owner = "mariacmartins"
repo = "computational_thinking_with_python"

url = f"https://api.github.com/repos/{owner}/{repo}"

resposta = requests.get(url)

if resposta.status_code == 200:
    dados = resposta.json()

    print("Nome:", dados["full_name"])
    print("Descrição:", dados["description"])
    print("Linguagem:", dados["language"])
    print("Stars:", dados["stargazers_count"])
    print("Forks:", dados["forks_count"])
    print("Issues abertas:", dados["open_issues_count"])
    print("Criado em:", dados["created_at"])
    print("Última atualização:", dados["updated_at"])
    print("URL:", dados["html_url"])
else:
    print("Repositório não encontrado.")
```

```python
# Endpoint de Commits
url = f"https://api.github.com/repos/{owner}/{repo}/commits"

resposta = requests.get(url)

commits = resposta.json()
```

```python
for commit in commits[:5]:
    print("SHA:", commit["sha"][:7])
    print("Mensagem:", commit["commit"]["message"])
    print("Autor:", commit["commit"]["author"]["name"])
    print("Data:", commit["commit"]["author"]["date"])
    print("-" * 50)
```

```python
# Endpoint de PRs
owner = "mariacmartins"
repo = "plpred"
url = f"https://api.github.com/repos/{owner}/{repo}/pulls"

params = {
    "state": "closed"
}

resposta = requests.get(
    url,
    params=params
)

pull_requests = resposta.json()

for pr in pull_requests:
    print(f"#{pr['number']} - {pr['title']}")
    print("Autor:", pr["user"]["login"])
    print("Criado em:", pr["created_at"])
    print("URL:", pr["html_url"])
    print("-" * 50)
```
