## Estruturas de Repetição - While

```python
# O Contador Básico
contador = 1

while contador <= 5:
    print(f"Número {contador}")
    contador += 1  # Crucial: sem isso, teremos um loop infinito!

print("Fim!")
```

```python
# Entendendo while com exercício de despertador!
resposta = "s"

while resposta == "s":
    resposta = input("Deseja dormir mais 5 minutos? (s/n): ")
    if resposta == "s":
      print('Mais 5min adicionados ao despertador!')

print("Bom dia! Hora de levantar.")
```

```python
# Tabuada
num = int(input("Qual tabuada você quer ver? "))
i = 1

while i <= 10:
    print(f"{num} * {i} = {num * i}")
    i += 1
```

```python
# Desafio das médias

# Define quantas vezes o loop irá rodar com base na entrada do usuário
total_numeros = int(input("Quantos números quer inserir? "))

# Inicializa as variáveis de controle (acumulador e contador)
soma = 0        # Variável que guardará a soma total dos números
contador = 0    # Variável que controla quantas vezes já pedimos um número

# O loop continua enquanto o contador for menor que o total solicitado
while contador < total_numeros:
    # Solicita um número e usa {contador + 1} para mostrar ao usuário
    # (ex: "Digite o 1º número" em vez de 0º)
    valor = float(input(f"Digite o {contador + 1}º número: "))

    # Adiciona o número digitado à nossa soma acumulada
    soma += valor

    # Incrementa o contador para avançar para a próxima repetição
    # Essencial para não entrar em loop infinito!
    contador += 1

# Após sair do while, calcula a média dividindo a soma pela quantidade
media = soma / total_numeros

# Exibe o resultado final formatado com 2 casas decimais (: .2f)
print(f"A média final é: {media:.2f}")
```

```python
# Desafio da adivinhação
from random import randint
computador = randint(0, 10)
tentativas = 0
print('---------JOGO DA ADIVINHAÇÃO----------')
print('Tente adivinhar qual número eu pensei!')

acertou = False
while not acertou:
  jogador = int(input('Digite um palpite: '))
  tentativas +=1
  if jogador == computador:
    acertou = True
print(f'Acertou com {tentativas} tentativas!')
```

```python
"""
DESAFIO DO ORÇAMENTO
"""
#1. Definimos o limite máximo que o usuário pode gastar.
# Convertemos para float porque dinheiro tem casas decimais.
orcamento = float(input("Qual o seu orçamento total? R$ "))

# 2. Inicializamos o acumulador. Ele começa em zero porque
# no início o usuário ainda não comprou nada.
total_gasto = 0

# 3. O loop WHILE verifica: "O que eu já gastei ainda cabe no meu orçamento?"
# Enquanto o total_gasto for menor ou igual ao orçamento, o bloco abaixo repete.
while total_gasto <= orcamento:
    # Pedimos o valor do produto atual.
    # Usamos uma f-string para mostrar o acumulador em tempo real.
    gasto = float(input(f"Valor do item (Total gasto: R$ {total_gasto:.2f}): "))

    # Somamos o novo gasto ao que já tínhamos acumulado.
    total_gasto += gasto

# 4. Quando o 'total_gasto' se torna MAIOR que o 'orcamento', o loop quebra.
# Agora calculamos a diferença para mostrar o prejuízo.
excedente = total_gasto - orcamento

# 5. Exibimos a mensagem final. O :.2f garante que apareça R$ 10.50 e não 10.5.
print(f"Orçamento estourado! Você gastou R$ {excedente:.2f} além do limite.")
```

### O Grande Desafio: A Viagem de Carro
(15min)

Enunciado:
Você está em um carro viajando para uma cidade que fica a 100 km de distância. O objetivo é chegar aos 100 km.

Comece com a distancia_percorrida = 0.

Enquanto a distância for menor que 100:

Pergunte: "Quantos quilômetros você percorreu nesta última hora?".

Some esse valor à distancia_percorrida.

Mostre: "Você já percorreu X km de 100 km".

Quando chegar a 100 (ou passar disso), o loop deve parar e mostrar: "Parabéns! Você chegou ao seu destino!".

```python
distancia_percorrida = 0

while distancia_percorrida < 100:
    trecho = float(input("Quantos quilômetros você percorreu nesta última hora? "))

    # Soma o que foi percorrido no trecho ao total
    distancia_percorrida = distancia_percorrida + trecho

    # Exibe o progresso
    print(f"Você já percorreu {distancia_percorrida} km de 100 km.")

print("Parabéns! Você chegou ao seu destino!")
```

### Comando break

O comando break é uma ferramenta de controle de fluxo fundamental no Python. Em tradução livre, significa "interromper".

Sua função é forçar a saída imediata de um loop (seja while ou for).

Por que usar o break?
Muitas vezes, a condição de parada de um while não é conhecida no início. Ela pode depender de uma interação do usuário ou de um erro que ocorre no meio do processamento.

```python
"""
Contexto:
Imagine que você está procurando um número específico dentro de uma sequência.
Assim que encontrar, não há motivo para continuar procurando.
"""

numero_secreto = 7
tentativa = 1

while tentativa <= 10:
    if tentativa == numero_secreto:
        print("Encontrei o número secreto!")
        break  # O programa para no 7, nem tenta o 8, 9 ou 10

    print(f"Testando o número {tentativa}")
    tentativa += 1
```

```python
contador = 1
while contador < 10:
    print(contador)
    contador += 1
    break # Isso aqui vai parar o loop logo na primeira volta!
```

```python
contador = 1
while contador < 10:
    if contador == 5:
        break # Só para se uma condição específica acontecer
    print(contador)
    contador += 1
```

```python
# Carrinho de Compras (E-commerce)
# Quando um cliente adiciona produtos ao carrinho, ele pode decidir parar de comprar a qualquer momento ao clicar em "Finalizar" ou "Sair".

total_compra = 0

while True:
    preco = float(input("Digite o preço do item (ou 0 para finalizar a compra): "))

    if preco == 0:
        # Aqui o break encerra o processo de compra
        break

    total_compra += preco
    print(f"Item adicionado. Total parcial: R$ {total_compra:.2f}")

print(f"Compra finalizada. Valor total a pagar: R$ {total_compra:.2f}")
```

### Comando continue
O continue é o "irmão" do break, mas ele não encerra o loop. Ele serve para pular uma rodada específica.



```python
numero = 0

while numero < 10:
    numero += 1

    if numero == 5:
        print("Pulando o número 5...")
        continue # O 'print' abaixo não será executado para o 5

    print(f"Contagem: {numero}")
```

```python
musica_atual = 1

print("--- PLAYER VIRTUAL ---")

while musica_atual <= 10:
    print(f"[Tocando agora: Música {musica_atual}]")

    # Interação com o usuário
    acao = input("Ouvir (o), Pular (p) ou Sair (s)?").lower()

    if acao == 's':
        print("Desligando o player...")
        break  # Interrompe tudo e sai do loop

    if acao == 'p':
        print(f"Pulando a música {musica_atual}...")
        musica_atual += 1
        continue  # Volta para o início do while para a PRÓXIMA música

    # Se escolheu 'o' (ou qualquer outra tecla), ele executa o que está abaixo
    print(f"Você ouviu a música {musica_atual} inteira.")
    musica_atual += 1

if musica_atual > 10:
    print("Playlist finalizada com sucesso!")
else:
    print("Player encerrado pelo usuário.")
```
