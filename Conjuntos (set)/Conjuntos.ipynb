# Principais métodos

```python
#add() - Adiciona elemento.
nomes = {"Ana"}

nomes.add("Carlos")

print(nomes)
```

```python
# update() - Adiciona vários.
nomes = {"Ana"}

nomes.update(["Pedro","Maria"])

print(nomes)
```

```python
# remoção de elementos
nomes = {'Paulo', 'Ana', 'Pedro', 'Maria'}
nomes.remove('Ana')
nomes.discard('Maria')
print(nomes)
```

### Desafio envolvendo vários conjuntos


Três departamentos de uma empresa enviaram listas de funcionários.

rh = {
    "Ana",
    "Carlos",
    "Pedro",
    "Lucas"
}

financeiro = {
    "Carlos",
    "Maria",
    "Lucas",
    "Fernanda"
}

ti = {
    "Lucas",
    "Fernanda",
    "Julia",
    "Pedro"
}

Encontre:
Funcionários que aparecem nos três departamentos.
Funcionários que aparecem em pelo menos dois departamentos.
Funcionários que aparecem em apenas um departamento.

```python
rh = {
    "Ana",
    "Carlos",
    "Pedro",
    "Lucas"
}

financeiro = {
    "Carlos",
    "Maria",
    "Lucas",
    "Fernanda"
}

ti = {
    "Lucas",
    "Fernanda",
    "Julia",
    "Pedro"
}
```

```python
# 1. Nos três departamentos
tres = rh & financeiro & ti
print("Três departamentos:", tres)
```

```python
# 2. Em pelo menos dois departamentos
dois_ou_mais = (
    (rh & financeiro)
    | (rh & ti)
    | (financeiro & ti)
)
print("Pelo menos dois:", dois_ou_mais)
```

```python
# 3. Apenas um departamento
todos = rh | financeiro | ti
apenas_um = todos - dois_ou_mais
print("Apenas um departamento:", apenas_um)
```
