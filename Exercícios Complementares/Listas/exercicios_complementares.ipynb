**Tratamento de dados em sequência de DNA**

Uma sequência genética veio com erros "X": dna = ["A", "T", "X", "G", "C", "X", "A"].

Descubra em qual posição está o primeiro erro "X".

Remova a primeira ocorrência do erro "X".

```python
# Resolução Tratamento de dados em sequência de DNA
dna = ["A", "T", "X", "G", "C", "X", "A"]

# 1. Localizando o índice
print(f"O primeiro erro está no índice: {dna.index('X')}")

# 2. Removendo o primeiro 'X'
dna.remove("X")
print(dna)
```

**Fluxo de Checkout**

Cenário: Você está desenvolvendo o carrinho de compras de um site.

A lista carrinho = ["Notebook", "Mouse", "Teclado"].

O cliente decidiu adicionar um "Monitor".

Ele percebeu que o "Mouse" era o modelo errado. Acesse o índice dele e substitua por "Mouse Gamer".

Por fim, o cliente desistiu do primeiro item ("Notebook"). Remova-o da lista.

```python
# Resolução: Fluxo de Checkout
carrinho = ["Notebook", "Mouse", "Teclado"]

# 1. Adicionando múltiplos itens (extend)
carrinho.append('Monitor')

# 2. Substituindo via índice (Mouse está no índice 1)
carrinho[1] = "Mouse Gamer"

# 3. Removendo o primeiro item (índice 0)
carrinho.pop(0)

print(f"Carrinho finalizado: {carrinho}")
```

## For

**Tratamento de dados de produtos**

Cenário: Você recebeu uma lista de códigos de produtos (SKUs) de um fornecedor. Os dados estão inconsistentes: alguns têm espaços extras, outros estão em letras minúsculas e há códigos duplicados que não podem ser cadastrados duas vezes no sistema.

lista_skus_brutos = [ " iphone-13-preto ", "GALAXY-S21-PRATA", "iphone-13-preto",
" macbook-m1-cinza ", "galaxy-s21-prata "
]

Remova espaços em branco e converta tudo para Letras Maiúsculas.
Garanta que a lista final não contenha códigos repetidos.

```python
#Resolução: Tratamento de dados de SKUs (produtos)

# Dados brutos vindos do fornecedor
lista_skus_brutos = [
    " iphone-13-preto ",
    "GALAXY-S21-PRATA",
    "iphone-13-preto",    # Duplicado
    "  macbook-m1-cinza ",
    "galaxy-s21-prata "   # Duplicado (em minúsculo)
]

# Lista para armazenar os dados tratados e únicos
lista_skus_limpos = []

for item in lista_skus_brutos:
    # 1. Tratamento: Remove espaços e coloca em MAIÚSCULO
    sku_tratado = item.strip().upper()

    # 2. Verificação de Duplicidade:
    # Só adicionamos se o SKU tratado ainda não estiver na lista final
    if sku_tratado not in lista_skus_limpos:
        lista_skus_limpos.append(sku_tratado)
        print(f"SKU Aceito: {sku_tratado}")
    else:
        print(f"SKU Duplicado Ignorado: {sku_tratado}")

print("\n--- Relatório Final de SKUs Únicos ---")
print(lista_skus_limpos)
```

**Sistema de descontos**

Contexto: Você está programando o caixa de uma papelaria. Para queimar o estoque, o dono da loja criou uma regra de desconto automática:

Produtos Acima de R$ 10,00: Ganham 10% de desconto.

Produtos Populares (Até R$ 10,00): Ganham 5% de desconto.

Lista a ser usada inicialmente:

precos_originais = [5.00, 15.00, 8.00, 20.00, 10.00]

O que o código deve fazer:

Percorrer a lista de preços originais.

Verificar o valor de cada item usando um if.

Aplicar o desconto correto e salvar o novo valor em uma lista chamada precos_com_desconto.

```python
## Resolução: Sistema de descontos
precos_originais = [5.00, 15.00, 8.00, 20.00, 10.00]

# Lista para guardar os novos valores calculados
precos_com_desconto = []

for preco in precos_originais:
    # Regra 1: Se for maior que 10 reais, desconto de 10% (multiplica por 0.90)
    if preco > 10:
        novo_valor = preco * 0.90

    # Regra 2: Caso contrário (menor ou igual a 10), desconto de 5% (multiplica por 0.95)
    else:
        novo_valor = preco * 0.95

    # Adiciona o resultado na lista final
    precos_com_desconto.append(novo_valor)

# Exibe as duas listas para comparação
print(f"Preços Originais: {precos_originais}")
print(f"Preços com Desconto: {precos_com_desconto}")
```
