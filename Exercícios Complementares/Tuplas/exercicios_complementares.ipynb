## Tuplas

**Menu Restaurante**

Crie uma tupla com as opções de bebidas de uma lanchonete: ("Suco", "Água", "Refrigerante"). Peça para o usuário digitar uma bebida e diga se a lanchonete possui essa opção ou não.

```python
cardapio = ("Suco", "Água", "Refrigerante")

pedido = input("O que deseja beber? ")

# Usando o 'in' para verificar a existência
if pedido in cardapio:
    print("Temos essa bebida no estoque!")
else:
    print("Desculpe, não temos essa opção.")
```

**Pontuação/Ranking**

Um jogador fez três partidas e você guardou as pontuações na tupla pontos = (120, 150, 90). Use comandos simples para mostrar a maior pontuação e a soma total de pontos.

```python
pontos = (120, 150, 90)

# Usando comandos diretos do Python
print("Sua maior pontuação foi:", max(pontos))
print("Total de pontos acumulados:", sum(pontos))
```
