**EXERCÍCIO 1:**

Preencha uma matriz 3x2 com números inteiros informados pelo usuário e exiba a matriz.

```python
# 1. Criamos a caixa principal (matriz) vazia
matriz = []

print("Preencha a matriz 3x2:")

# 2. O primeiro laço controla as LINHAS (são 3)
for i in range(3):
    linha = [] # Criamos uma linha nova e vazia para cada andar

    # 3. O segundo laço controla as COLUNAS (são 2)
    for j in range(2):
        # Pedimos o número e já convertemos para inteiro (int)
        numero = int(input(f"Digite o valor para [{i}][{j}]: "))

        # Colocamos o número dentro da linha atual
        linha.append(numero)

    # 4. Agora que a linha tem 2 números, "fechamos" ela na matriz
    matriz.append(linha)

# 5. Exibimos a matriz de forma organizada
print("\nMatriz preenchida:")
for linha in matriz:
    print(linha)
```

**EXERCÍCIO 2**

Preencha uma matriz 5x5 com números aleatórios e exiba a matriz.

```python
import random

# 1. Criar e preencher a matriz 5x5
matriz = []
for i in range(5):
    linha = []
    for j in range(5):
        n = random.randint(1, 100) # Sorteia números entre 1 e 100
        linha.append(n)
    matriz.append(linha)

# 2. Exibir a matriz de forma organizada
print("Matriz Gerada:")
for linha in matriz:
    print(linha)
```

**EXERCÍCIO 3**
Use a matriz gerada no exercício 2 e multiplique todos os elementos por 2.

```python
# Supondo que 'matriz' já existe e está preenchida...

# 1. Percorrendo a matriz para multiplicar cada valor por 2
for linha in matriz:
    for i in range(len(linha)):
        linha[i] *= 2

# 2. Exibindo a matriz resultante de forma organizada
print("--- Matriz Multiplicada por 2 ---")
print(matriz)
```
